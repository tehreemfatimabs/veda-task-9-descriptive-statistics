This project analyzes numerical variables from the Titanic passenger dataset using Python and Pandas in Jupyter Notebook. The selected variables are Age, Fare, SibSp (siblings and spouses aboard), and Parch (parents and children aboard).

This project demonstrates how descriptive statistics can transform raw data into meaningful insights and support better data interpretation and decision-making.

# Load Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

# Load Dataset

```python
df = pd.read_csv("titanic.csv")
```

```python
# Select  Numeric Columns for Analysis
columns = ["age","fare","sibsp","parch"]
```

# Basic EDA

```python
print("Datset Shape",df.shape)
print("\n Dataset First five Rows")
print(df[columns].head())
print("\nMissing Values:")
print(df[columns].isnull().sum())
```

# Descriptive Statistics

```python
summary = df[columns].describe().T
summary["mode"] = df[columns].mode().iloc[0]
summary = summary.round(2)
summary
```

# Interpretation of Descriptive Statistics

The descriptive statistical analysis of the Titanic dataset reveals differences in the central tendency, variability, and distribution of passenger age, ticket fare, and family-related variables.

### 1. Age

The Age column contains 714 non-missing observations, with a mean of 29.70 years and a median of 28.00 years. Since the mean is slightly greater than the median, the distribution may be positively (right) skewed, indicating that some older passengers increase the average age. The standard deviation of 14.53 years indicates considerable variation in passenger ages. The middle 50% of observed ages lie approximately between 20.12 and 38.00 years.

### 2. Fare

The Fare column contains 891 observations, with a mean of 32.20 and a median of 14.45. The substantial difference between these values, together with a standard deviation of 49.69, suggests a strongly right-skewed distribution. A relatively small number of passengers paid considerably higher fares, increasing both the mean and the variability. The middle 50% of fares range from approximately 7.91 to 31.00, showing that most passengers paid substantially less than the maximum fare of 512.33.

### 3. SibSp

The SibSp variable has a mean of 0.52 and a median of 0.00. This indicates that passengers generally travelled without a sibling or spouse aboard, although some travelled with one or more family members. The standard deviation of 1.10 indicates variation in family accompaniment. The distribution is non-negative and concentrated at zero, so a normal distribution should not be assumed.

### 4. Parch

The Parch variable has a mean of 0.38 and a median of 0.00. This suggests that most passengers had no parent or child aboard, while a smaller proportion travelled with family members. The standard deviation of 0.81 indicates relatively limited absolute variation compared with the Fare variable. As with SibSp, the data are discrete counts and are concentrated at zero rather than being normally distributed.

### Overall Conclusion

The analysis indicates that passenger fares are more variable and likely more strongly right-skewed than passenger ages. The SibSp and Parch variables are concentrated at zero, reflecting the fact that many passengers did not have the corresponding family members aboard. Comparing means and medians provides an initial indication of skewness, while standard deviations and percentiles help explain variability and the spread of the central 50% of observations. Distribution plots would provide further evidence when assessing skewness and normality.

# Visualizations

## Interview Questions
### Question 1: When is the median a better measure of center than the mean?

Answer:

The median is a better measure of center when a dataset contains extreme values or is highly skewed. Unlike the mean, the median is resistant to outliers and is not greatly affected by unusually large or small observations.

For example, in the Titanic dataset, some passengers paid exceptionally high ticket fares. These values can increase the mean fare, while the median provides a more robust representation of the typical passenger's fare.

### Question 2: What does a large standard deviation tell you about a dataset?

Answer:

A large standard deviation indicates that observations are widely dispersed around the mean, suggesting greater variability in the data. A small standard deviation indicates that observations are generally closer to the mean.

However, standard deviation should be interpreted alongside the mean and the variable's scale. A large standard deviation alone does not indicate whether the data is normally distributed.

### Question 3: What is the difference between the mean and the median, and how can they help identify skewness?

Answer:

The mean is the sum of all observations divided by the number of valid observations, whereas the median is the middle value when observations are arranged in order.

When the mean is greater than the median, the distribution may be positively (right) skewed. When the mean is less than the median, it may be negatively (left) skewed. If the two values are approximately equal, the distribution may be roughly symmetric. These comparisons provide useful clues, but a histogram or other distributional analysis is needed to confirm the shape.
